```python
from pyspark.sql import DataFrame
from pyspark.sql.types import StructType

def read_bronze_csv(filepath: str, schema: StructType, sample_rows: int = None) -> DataFrame:
    """Read a CSV with an explicit schema. Optionally limit rows for dev/testing."""
    reader = (
        spark.read
        .format("csv")
        .option("header", True)
        .option("delimiter", ",")
        .option("escape", '"')
        .option("multiLine", True)
        .schema(schema)
    )
    df = reader.load(filepath)
    if sample_rows:
        df = df.limit(sample_rows)
    return df
```
