Title: Health Insurance Marketplace Analytics Platform

Business Case: Explore health and dental plans data in the US Health Insurance Marketplace.
The marketplace generates enormous volumes of plan data for rates, benefits, networks, and coverage areas across all US
states and multiple plan years. The data is available in flat CSVs with no repeatable pipeline, no versioned transforms, and no
systematic way to answer policy or consumer questions at scale.

This project builds a structured ETL and analytics platform for transforming & build some of the below business use cases.

Question to being resolved:
- How do plan rates and benefits vary across states?
- How do plan benefits relate to plan rates?
- How do plan rates vary by age?
- How do plans vary across insurance network providers?


Data links:
https://www.kaggle.com/datasets/hhs/health-insurance-marketplace

```python
import kagglehub
from kagglehub import KaggleDatasetAdapter

benefits_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "BenefitsCostSharing.csv"
)
print("First 5 records:", benefits_df.head())

business_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "BusinessRules.csv"
)
print("First 5 records:", business_df.head())

crosswalk2015_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "Crosswalk2015.csv"
)
print("First 5 records:", crosswalk2015_df.head())

crosswalk2016_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "Crosswalk2016.csv"
)
print("First 5 records:", crosswalk2016_df.head())

network_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "Network.csv"
)
print("First 5 records:", network_df.head())

plan_attributes_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "PlanAttributes.csv"
)
print("First 5 records:", plan_attributes_df.head())

rate_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "Rate.csv"
)
print("First 5 records:", rate_df.head())

service_area_df = kagglehub.load_dataset(
  KaggleDatasetAdapter.PANDAS,
  "hhs/health-insurance-marketplace",
  "ServiceArea.csv"
)
print("First 5 records:", service_area_df.head())
```

```python
print("Benefits Cost Sharing: \n", benefits_df.columns)
print("Business Rules: \n", business_df.columns)
print("Crosswalk 2015: \n", crosswalk2015_df.columns)
print("Crosswalk 2016: \n", crosswalk2016_df.columns)
print("Network: \n", network_df.columns)
print("Plan Attributes: \n", plan_attributes_df.columns)
print("Rate: \n", rate_df.columns)
print("ServiceArea: \n", service_area_df.columns)
```
